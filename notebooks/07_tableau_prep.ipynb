```python
year_check = pd.read_sql("""
    SELECT year, COUNT(DISTINCT round) as races_count, MAX(race_date) as last_race
    FROM dim_race
    WHERE year >= 2023
    GROUP BY year
    ORDER BY year
""", engine)
year_check
```

```python
leader_check = pd.read_sql("""
    SELECT d.forename, d.surname, SUM(f.points) as total_points
    FROM fact_race_results f
    JOIN dim_driver d ON f.driver_key = d.driver_key
    JOIN dim_race r ON f.race_key = r.race_key
    WHERE r.year = 2024
    GROUP BY d.forename, d.surname
    ORDER BY total_points DESC
    LIMIT 5
""", engine)
leader_check
```
